```python
# Для 
import os
import pandas as pd
import cv2

def parse_icdar_gt(txt_path, images_dir):
    data = []
    with open(txt_path, 'r', encoding='utf-8-sig') as f:  # Используем 'utf-8-sig' для удаления BOM
        lines = f.readlines()
    
    for line in lines:
        line = line.strip()
        if not line:
            continue
        
        parts = line.split(',')
        if len(parts) < 9:
            continue
        
        try:
            coords = list(map(int, parts[:8]))  # Теперь BOM не мешает
            text = ','.join(parts[8:])
            
            x_coords = coords[::2]
            y_coords = coords[1::2]
            bbox = [min(x_coords), min(y_coords), max(x_coords), max(y_coords)]
            
            image_id = os.path.basename(txt_path).replace('gt_', '').replace('.txt', '.jpg')
            image_path = os.path.join(images_dir, image_id)
            
            if not os.path.exists(image_path):
                print(f"⚠️ Изображение не найдено: {image_path}")
                continue
            
            data.append({
                'image_id': image_id,
                'label': text,
                'bbox': bbox,
                'image': image_path
            })
        except ValueError as e:
            print(f"Ошибка в строке: {line}\nОшибка: {e}")
            continue
    
    return pd.DataFrame(data)

# Пути к данным (относительно расположения скрипта)
base_dir = 'Dataset1'

# Тестовые данные
test_gt_dir = os.path.join(base_dir, 'test', 'Challenge4_Test_Task1_GT')
test_images_dir = os.path.join(base_dir, 'test', 'ch4_test_images')

# Тренировочные данные
train_gt_dir = os.path.join(base_dir, 'train', 'ch4_training_localization_transcription_gt')
train_images_dir = os.path.join(base_dir, 'train', 'ch4_training_images')

# Загрузка всех данных
def load_all_data(gt_dir, images_dir):
    all_data = []
    for txt_file in os.listdir(gt_dir):
        if txt_file.endswith('.txt'):
            txt_path = os.path.join(gt_dir, txt_file)
            df = parse_icdar_gt(txt_path, images_dir)
            all_data.append(df)
    return pd.concat(all_data, ignore_index=True)


test_df_1 = load_all_data(test_gt_dir, test_images_dir)
train_df_1 = load_all_data(train_gt_dir, train_images_dir)


final_df_1 = pd.concat([test_df_1, train_df_1], ignore_index=True)
'''
# Проверка результата
print("Тестовые данные:")
print(test_df.head())
print("\nТренировочные данные:")
print(train_df.head()) '''
```
