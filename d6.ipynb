```python
import xml.etree.ElementTree as ET
import pandas as pd
import os

def parse_cvat_annotations(xml_path, images_dir):
    """
    Парсит CVAT XML аннотации в DataFrame
    
    Args:
        xml_path (str): Путь к XML файлу аннотаций
        images_dir (str): Путь к папке с изображениями
        
    Returns:
        pd.DataFrame: DataFrame с колонками ['img_id', 'label', 'bbox', 'Image_path']
    """
    tree = ET.parse(xml_path)
    root = tree.getroot()
    
    data = []
    
    for image in root.findall('image'):
        # Получаем информацию об изображении
        image_name = image.get('name')
        img_id = os.path.splitext(os.path.basename(image_name))[0]
        image_path = os.path.join(images_dir, os.path.basename(image_name))
        
        # Парсим все bounding boxes
        for box in image.findall('box'):
            label = box.get('label')
            text = box.find('attribute').text if box.find('attribute') is not None else ''
            
            # Формируем bbox в формате [[xmin, ymin], [xmax, ymax]]
            bbox = [
                [float(box.get('xtl')), float(box.get('ytl'))],
                [float(box.get('xbr')), float(box.get('ybr'))]
            ]
            
            data.append({
                'img_id': img_id,
                'label': text,  # Используем текст из атрибута как метку
                'bbox': bbox,
                'Image_path': image_path
            })
    
    return pd.DataFrame(data, columns=['img_id', 'label', 'bbox', 'Image_path'])

# Пути к данным
xml_path = 'Dataset6/archive/annotations.xml'
images_dir = 'Dataset6/archive/images'

# Парсим аннотации
df = parse_cvat_annotations(xml_path, images_dir)

# Проверяем результат
print(f"Всего аннотаций: {len(df)}")
print(df.head())

# Сохраняем в CSV
df.to_csv('d6.csv', index=False)
```

```python
import xml.etree.ElementTree as ET
import pandas as pd
import os
from sklearn.model_selection import train_test_split

def parse_cvat_annotations(xml_path, images_dir):
    """
    Парсит CVAT XML аннотации в DataFrame
    
    Args:
        xml_path (str): Путь к XML файлу аннотаций
        images_dir (str): Путь к папке с изображениями
        
    Returns:
        pd.DataFrame: DataFrame с колонками ['img_id', 'label', 'bbox', 'Image_path']
    """
    tree = ET.parse(xml_path)
    root = tree.getroot()
    
    data = []
    
    for image in root.findall('image'):
        # Получаем информацию об изображении
        image_name = image.get('name')
        img_id = os.path.splitext(os.path.basename(image_name))[0]
        image_path = os.path.join(images_dir, os.path.basename(image_name))
        
        # Парсим все bounding boxes
        for box in image.findall('box'):
            label = box.get('label')
            text = box.find('attribute').text if box.find('attribute') is not None else ''
            
            # Формируем bbox в формате [[xmin, ymin], [xmax, ymax]]
            bbox = [
                [float(box.get('xtl')), float(box.get('ytl'))],
                [float(box.get('xbr')), float(box.get('ybr'))]
            ]
            
            data.append({
                'img_id': img_id,
                'label': text,  # Используем текст из атрибута как метку
                'bbox': bbox,
                'Image_path': image_path
            })
    
    return pd.DataFrame(data, columns=['img_id', 'label', 'bbox', 'Image_path'])

# Пути к данным
xml_path = 'Dataset6/archive/annotations.xml'
images_dir = 'Dataset6/archive/images'

# Парсим аннотации
df = parse_cvat_annotations(xml_path, images_dir)

# Проверяем результат
print(f"Всего аннотаций: {len(df)}")
print(df.head())

# Разделяем на train и test в соотношении 5:1
train_df, test_df = train_test_split(df, test_size=0.1667, random_state=42)

# Проверяем размеры
print(f"Train annotations: {len(train_df)}")
print(f"Test annotations: {len(test_df)}")

# Сохраняем в CSV
train_df.to_csv('d6_train.csv', index=False)
test_df.to_csv('d6_test.csv', index=False)

print("Train dataset saved to 'd6_train.csv'")
print("Test dataset saved to 'd6_test.csv'")
```
